## print()

This function can print (display) various objects.

Here are some easy example:

````python
greet = 'Welcome to'
department = 'Econ'
course = 57003
instructor = 'Arie Beresteanu'
credit = True
````

````python
print(greet,department,course)
print('Instructor:',instructor)
print('For credit:',credit)
````

**Notice \#1:** ```print()``` adds blank spaces between the strings without you having to do so!

**Notice \#2:** ```print()``` can accept an arbitrary number of inputs of different kind

**Notice \#3:** Every ```print()``` command appears in a new line 

What if we do not want python to add spaces? For example, when printing a URL (web address).

````python
baseURL = 'https:\\sites.pitt.edu'
user = 'arie'
````

````python
print(baseURL,r'\~',user)  #bad
````

There are **two** problems here. First, we have undesired spaces. Second, we have a missing \ after the ```https:```. Let's fix the first problem first.

In python ```string1 + string2``` will yield a new string that is a concatenation of the two strings.

So insted of seperating the strings by commas, we sum them:

````python
print(baseURL+r'\~'+user)  #better
````

We still need to solve the missing ```\``` after the https. 

In Python and many other languages, the symbol ```\``` is called an _escape character_. What it means is that it indicates that the following character needs to be interpreted in a different way (usually literaly) than what the usual syntax may suggest. 

This concept is not easy to understand at first. A general rule is that if a character is missing in your print output (in our case a ```\```), you should _escape_ it by adding a ```\``` in front of it. 

Here is a fix:

````python
baseURL = r'https:\\sites.pitt.edu'  
user = 'arie'
````

````python
print(baseURL+r'\~'+user)  #best
````

By the way. the ```~``` in the middle didn't give us any problems because it doesn't have any special meaning. So, it's hard to give a precise rule, it's more a trial and error kind of process (at least for me). Just be mindful that sometimes characters don't show and they need to be _escaped_.

````python
print('It\'s not the best explanation, but it will do') #notice the \ there escaping the contraction apostrophe
````

### Formating ```print()``` statements

In general formating is written as follows:
```Python
f'{object:,format}'
```

```f'{ }'``` indicates that the object (e.g. number) inside the curely brackets is going to be formated. The object in our case is going to be a number or a variable containing a number. ```:,``` are sperators between the object and the format. ```format``` is some kind of instructions to operate on the object in order to give it a nicer format.


Comma seperators for 1000s and number of digits after the decimal point:

````python
prod21 = revenue21/employees21  #for convenience

print('Average labor productivity in 2021 is', f'{prod21:,.2f}')
````

````python
price = 12.5
year = 2023
country = "Israel"
print(f'the price was {price} in {year} in {country}')
````

**Notice \#1:** The above formating works only on ```float```s (real numbers). If applied to an integer you will get an error. So ```.2f``` stands for a float number with 2 digits after the decimal point and add 1000s commas.

**Notice \#2:** For integers try: ```print(f'{course:,d}')```.

**Notice \#3:** The above can be done also using the function ```format()```.

**Notice \#4:** To get a short help on a function in your notebook you can use a question mark. For example, if, in a new cell you'll type ```?format```, you will get the syntax of this function.

### ```sorted()```

sorting lists or tuples.

````python
profs = ['Ripoll', 'Vesterlund', 'Beresteanu','Rigotti','Wilson','Walsh']
sorted(profs)                #ascending order
````

````python
sorted(profs,reverse=True)  #descending order
````

````python
scores = (84,71,95,100,83)
scores_asc = sorted(scores)  #by default sorted() turns a tuple to a list 
print(scores_asc)
````

````python
scores_asc = tuple(sorted(scores))  #if you want to keep it a tuple
print(scores_asc)
````

````python
# note that the original list is not modified!
print(scores)
````
