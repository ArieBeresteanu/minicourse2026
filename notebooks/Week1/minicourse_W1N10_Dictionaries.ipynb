## Introduction
A dictionary in Python is a collection of key-value pairs. It is similar to a real-life dictionary where you have words (keys) and their definitions (values).

````python
person = ("Arie","Pittsburgh",19)
````

````python
person[0]
````

## Creating a Dictionary
An empty dictionary can be created using curly braces {} or the dict() function.

````python
# Creating an empty dictionary
empty_dict = {}
print("Empty Dictionary:", empty_dict)
````

Note: an alternative syntax is:

```Python
empty_dict = dict()
```

````python
# Creating a dictionary with initial key-value pairs
student_grades = {"Alice": 85,
                  "Bob": 90,
                  "Charlie": 78}

print("Dictionary with initial key-value pairs:", student_grades)

#Note: indentation within the definition of a dictionary is not necessary but helpful
````

## Accessing Dictionary Elements
You can access dictionary values using their corresponding keys.

````python
# Accessing values using keys
print("Alice's grade:", student_grades["Alice"])
````

````python
# ANother way to write the above

print(f'Alice\'s grade: {student_grades["Alice"]}')
````

## Modifying Dictionaries
You can add, update, or remove key-value pairs in a dictionary. In other words, dictionaries are mutable objects.

````python
# Adding a new key-value pair
student_grades["David"] = 92
print("After adding David:", student_grades)
````

````python
# Updating an existing value
student_grades["Alice"] = 88
print("After updating Alice's grade:", student_grades)
````

````python
# Removing a key-value pair using del
del student_grades["Charlie"]
print("After removing Charlie:", student_grades)
````

````python
# Removing a key-value pair using pop
bob_grade = student_grades.pop("Bob")
print("After removing Bob:", student_grades)
print("Bob's grade (popped value):", bob_grade)
````

## Dictionary Methods
Some common dictionary methods are `keys()`, `values()`, `items()`, `get()`, `clear()` `copy()`, and `popitem()`.

````python
# Demonstrating keys(), values(), and items()
print("Keys:", student_grades.keys())
print("Values:", student_grades.values())
print("Items:", student_grades.items())
````

````python
# Using get() method
print("David's grade using get():", student_grades.get("David"))
````

````python
student_grades['Alice']
````

````python
list(student_grades.values())
````

## Looping through Dictionaries
You can loop through keys, values, and key-value pairs of a dictionary.

````python
# Looping through keys
print("Keys in the dictionary:")
for key in student_grades:
    print(key)
````

````python
# Looping through values
print("Values in the dictionary:")
for value in student_grades.values():
    print(value)
````

````python
# Looping through key-value pairs
print("Key-value pairs in the dictionary:")
for key, value in student_grades.items():
    print(key, ":", value)
````

## Dictionary Comprehensions
Dictionary comprehensions provide a concise way to create dictionaries.

````python
# Example of a dictionary comprehension
squares = {x: x*x for x in range(6)}
print("Squares dictionary:", squares)
````

# Nested Dictionaries
Dictionaries can contain other dictionaries as values.

````python
# Example of a nested dictionary
nested_dict = {
    "class1": {"Alice": 85, "Bob": 90},
    "class2": {"Charlie": 78, "David": 92}
}
print("Nested Dictionary:", nested_dict)
````

### Pretty print

We need to use a package called `json`.

## Practical Examples
Let's see some practical examples

````python
# Example 1: Counting the frequency of elements in a list

print("Example 1: Counting the frequency of elements in a list")
example_list = ["Employed", "Retired", "Employed", "Unemployed", "Employed", "Retired", "Employed"]

frequency = {}
for item in example_list:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
print("Frequency of elements:", frequency)
````

````python
# Example 2: Creating a simple contact book
print("Example 2: Creating a simple contact book")

contacts = {
    "John Doe": {"phone": "123-456-7890", "email": "john@example.com"},
    "Jane Smith": {"phone": "987-654-3210", "email": "jane@example.com"}
}
print("Contact book:", contacts)
````

````python
# Accessing nested dictionary
print("John Doe's phone number:", contacts["John Doe"]["phone"])
````
