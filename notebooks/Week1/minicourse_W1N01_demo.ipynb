````python
def presentValue(payment,periods,rate):
    # This function computes the present value of a future payment
    R = 1 + rate
    pv = payment/(R**periods)
    return pv
````

**ALL** inputs to the function **must** be included in the **right order**.

````python
presentValue(100,3,0.05)
````

````python
# Uncomment the line below. Will it work?

# presentValue(100,3)
````

Order of arguments matter! Even if the function will run (syntax is correct), the answer is wrong!

````python
presentValue(100,0.05,3)
````

**Note:** The function `presentValue()` requiers exactly three positional arguments (inputs).

### An `if` `else` code block

````python
def collatz(n):
    if n%2 == 0:  # checking whether n is even or odd
        n = int(n/2)  #divide by 2
    else:
        n = 3*n + 1
    return n
````

**Question:** In n is even (i.e. divides by 2 without a remainder), why do we need to write `int(n/2)`?

**Answer:** `int()` turns a `float` into an integer. The result of `10/2` is `5.0` (a float) which is different  than `5` (an integer).

````python
# trying the function:

print(collatz(17))
````

````python
# We can assign the result of a function to another variable

y = collatz(17)
print(y)
````

````python
# Will this while loop end after a finite number of steps for every starting value of n?

n = 157
print(n) # ptinting the starting point

while n>1:  #as long as we didn't get back to 1
    n = collatz(n)
    print(n)
````
