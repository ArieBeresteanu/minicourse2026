## Present value

Let $P$ be a paymnet received $t$ periods in the future. Let $r$ be the discount rate (e.g. interest rate) used to calculate the present value. For simplicity, we assume that $t$ is in whole years and $r$ is the annual interest rate (as a decimal). 

The present value is then,
$$PV = \frac{P}{(1+r)^t}.$$

Next we write a Python function that computes the present value.

````python
def present_value_v1(P,r,t):
    return P / ((1 + r)**t)
````

````python
present_value_v1(100,0.05,3)
````

The fuction `present_value_v1` acceps three inputs: (P,r,t). The following conditions must be satisfied:
- The function is calle with exactly three inputs
- The inputs must appear in the right order

The following calls will produce an error:
```Python
present_value_v1(100,0.05)
present_value_v1(100,0.05,2,False)
```

The following calls will produce an nonsensical answer:
```Python
present_value_v1(0.05,100,2)
present_value_v1(100,-0.05,2)
```

**Note:** Calling the function with `present_value_v1(100,0.05,-2)` actually has a meaning. What is it?

Let's start by checking the validity of the inputs.
1. The future payment, `P`, should be a positive number, either a `float` or an `int`.
2. The interest rate, `r`, should be a positive `float` of value less than `1` (not a must, but it will make more sense)
3. The time period, `t`, should be a `float` or an `int` (we can allow it to be negative).

````python
def present_value_v1_1(P,r,t):
    #checking the input P
    if not (isinstance(P, (int, float)) and P > 0):
        print("Error: P must be a positive integer or float.")
        return None
    
    if not(isinstance(r,(float)) and r>0 and r<1):
        print("Error: r must be a positive float between 0 and 1")
        return None
    
    if not(isinstance(t,(int,float)) and t>0):
        print("Error: t must be a positive integer or a float")
        return None
    
    return P / ((1 + r)**t)
````

````python
present_value_v1_1(100,1.2,3)
````

## Default values

The above functionrequiers all three inputs. Suppose that $t=1$ is the default value. We would like to call the function with only the first two arguments.

````python
def present_value_v2(P,r,t=1):
    """
    Calculate the present value of a future payment.

    param P: float or int - The future payment, must be a positive number.
    param r: float - The annual interest rate (as a decimal).
    param t: float - (optional) The time in years until the payment is received. Default t=1.

    return: float - The present value, or None if inputs are of the wrong type or value.
    """

    
    #checking the input P
    if not (isinstance(P, (int, float)) and P > 0):
        print("Error: P must be a positive integer or float.")
        return None
    
    if not(isinstance(r,(float)) and r>0 and r<1):
        print("Error: r must be a positive float between 0 and 1")
        return None
    
    if not(isinstance(t,(int,float))):
        print("Error: t must be an integer or a float")
        return None
    
    return P / ((1 + r)**t) 
````

## Keyword parameters

So far each input to the function had to appear in the right order when the function was called. Suppose we want to allow the user to enter some of the inputs at an arbitrary order. In this case, the user will have to name the parameters provided to the function. 

To see this, let's first add some functionality to our present value function.

### Compounded interest

When dealing with compounded interest, the formula for calculating the present value of a future payment changes based on the compounding frequency. Let's say the interest is compounded $n$ times per year. The formula for the present value (PV) is then:

$$PV = \frac{P}{\left(1 + \frac{r}{n}\right)^{nt}}$$

Here:
- $ PV $ is the present value.
- $ P $ is the future payment.
- $ r $ is the annual interest rate.
- $ n $ is the number of times the interest is compounded per year.
- $ t $ is the number of years until the payment is received.

In this scenario, the interest rate $\frac{r}{n}$ is the rate per compounding period, and $nt$ is the total number of compounding periods over the time frame.

If $t$ is given in days, you need to convert it into years since $r$ is an annual rate. Assuming a standard year of 365 days, the formula becomes:

$$PV = \frac{P}{\left(1 + \frac{r}{n}\right)^{n\left(\frac{t}{365}\right)}}$$

This adjustment ensures that the compounding periods are appropriately accounted for over the actual time frame until the payment is received. The choice of $n$ depends on the compounding frequency (e.g., annually, semi-annually, quarterly, monthly, daily, etc.).

First step, Let's first add the additional functionality:

````python
def present_value_v2_1(P, r, t=1, compounded=False, n=1):
    """
    Calculate the present value of a future payment.

    :param P: float - The future payment.
    :param r: float - The annual interest rate (as a decimal).
    :param t: float - The time in years until the payment is received.
    :param compounded: bool - True if interest is compounded, False otherwise.
    :param n: int - Number of times interest is compounded per year (ignored if compounded is False).
    :return: float - The present value.
    """
    #checking the input P
    if not (isinstance(P, (int, float)) and P > 0):
        print("Error: P must be a positive integer or float.")
        return None
    
    if not(isinstance(r,(float)) and r>0 and r<1):
        print("Error: r must be a positive float between 0 and 1")
        return None
    
    if not(isinstance(t,(int,float))):
        print("Error: t must be an integer or a float")
        return None
   
    # computing present value:
    if compounded:
        return P / ((1 + r / n)**(n * t))
    else:
        return P / ((1 + r)**t)

````

````python
# Examples: 

PV = present_value_v2_1(100, 0.05, 1)  # Non-compounded interest
print(f'Without compounding: {PV}')

PV_compounded = present_value_v2_1(100, 0.05, 1, True, 12)  # Compounded monthly
print(f'With compounding, n=12: {PV_compounded}')
````

Now that we have the functionality, we can make the last two imputs to be keyword inputs.

````python
def present_value_v2_2(P, r, t, *, compounded=False, n=1):
    """
    Calculate the present value of a future payment.

    :param P: float - The future payment.
    :param r: float - The annual interest rate (as a decimal).
    :param t: float - The time in years until the payment is received.
    :param compounded: bool - True if interest is compounded, False otherwise (keyword-only argument).
    :param n: int - Number of times interest is compounded per year (ignored if compounded is False).
    :return: float - The present value.
    """
    #checking the input P
    if not (isinstance(P, (int, float)) and P > 0):
        print("Error: P must be a positive integer or float.")
        return None
    
    if not(isinstance(r,(float)) and r>0 and r<1):
        print("Error: r must be a positive float between 0 and 1")
        return None
    
    if not(isinstance(t,(int,float))):
        print("Error: t must be an integer or a float")
        return None
   
    # computing present value:
    if compounded:
        return P / ((1 + r / n)**(n * t))
    else:
        return P / ((1 + r)**t)

````

````python
# Example usage
PV = present_value_v2_2(100, 0.05, 1)  # Non-compounded interest
print(f'Without compounding: {PV}')

PV_compounded = present_value_v2_2(100, 0.05, 1, compounded=True, n=12)  # Compounded monthly
print(f'With compounding, n=12: {PV_compounded}')
````

````python
PV_compounded = present_value_v2_2(100, 0.05, 1, n=12, compounded=True)  # Compounded monthly
print(f'With compounding, n=12: {PV_compounded}')
````

In this version of the function, compounded is a keyword-only argument, which means it can only be specified by its name when calling the function. This is achieved by placing an asterisk `*` before compounded in the function definition. The asterisk `*` in the parameter list indicates that all following parameters are keyword-only and must be specified by their name. This can make the function calls clearer and prevent errors from passing arguments in the wrong order.

````python
def present_value_v3(P, r, t, *, compounded=False, n=1):
    """
    Calculate the present value of a future payment.

    :param P: float - The future payment.
    :param r: float - The annual interest rate (as a decimal).
    :param t: float - The time in years until the payment is received.
    :param compounded: bool - True if interest is compounded, False otherwise (keyword-only argument).
    :param n: int - Number of times interest is compounded per year (ignored if compounded is False).
    :return: float - The present value.
    """
    #checking the input P
    if not (isinstance(P, (int, float)) and P > 0):
        print("Error: P must be a positive integer or float.")
        return None
    
    if not(isinstance(r,(float)) and r>0 and r<1):
        print("Error: r must be a positive float between 0 and 1")
        return None
    
    if not(isinstance(t,(int,float))):
        print("Error: t must be an integer or a float")
        return None
   
    # computing present value:
    if compounded:
        return P / ((1 + r / n)**(n * t))
    else:
        return P / ((1 + r)**t)

````

## Clean up and final version

We now have the final version of the function that we can share. We can now drop the version number from its name.

````python
def present_value(*,Payment, rate, timePeriods, compounded=False, n=1):
    """
    Calculate the present value of a future payment.

    :param Payment: float - The future payment.
    :param rate: float - The annual interest rate (as a decimal).
    :param timePeriods: float - The time in years until the payment is received.
    :param compounded: bool - True if interest is compounded, False otherwise (keyword-only argument).
    :param n: int - Number of times interest is compounded per year (ignored if compounded is False).
    :return: float - The present value.
    """
    #checking the input Payment
    if not (isinstance(Payment, (int, float)) and Payment > 0):
        print("Error: Payment must be a positive integer or float.")
        return None
    
    if not(isinstance(rate,(float)) and rate>0 and rate<1):
        print("Error: rate must be a positive float between 0 and 1")
        return None
    
    if not(isinstance(timePeriods,(int,float))):
        print("Error: t must be an integer or a float")
        return None
   
    # computing present value:
    if compounded:
        return Payment / ((1 + rate / n)**(n * timePeriods))
    else:
        return Payment / ((1 + rate)**timePeriods)

````
