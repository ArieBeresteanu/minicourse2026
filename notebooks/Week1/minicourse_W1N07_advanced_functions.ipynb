## ```map()```

The ```map()``` function take a sequence (e.g. a tuple or a list) of objects and applies a certain function to each item in the sequence.

We saw the function ```min()``` that operates on a tuple or list. But the ```min()``` function takes the whole tuple or list and returns one number, the minimum. For example ```min([2.0, -1.2, 4.5])``` will return ```-1.2```.

What if we want to take the absolute value of each element in the list? In other words, we want to apply the ```abs()``` function to the list element by element. 

We could write a loop but that would be tedious.

Instead we can "map" the ```abs()``` function to the list element by element.

````python
myList = [2.0, -1.2, 4.5]
````

````python
# Uncomment the command below and run this cell. Will it work?

#abs(myList)
````

````python
mapList = map(abs,myList)
print(mapList)
````

````python
type(mapList)
````

As you can see, Python was being a bit too literal. The variable ```newList``` contains the operation of mapping the function ```abs()``` to the list ```myList``` rather than the _result_ of this mapping. 

We need to do something more to get the result of this mapping.

````python
newList = list(mapList)
print(newList)
````

````python
l = ["a","b","Jerusalem","HUJI"]
l_len =  list(map(len,l))
print(l_len)
````

**Explanation:**

Suppose we want to take each element of a list and check whether it is positive. If it is, we return the number itself. Otherwise we return zero.
$$
\begin{equation*}
lowerTrim(x) = \left\{
    \begin{array}\\
        x & \mbox{if } \ x >0 \\
        0 & \mbox{if } \ x \leq 0 
    \end{array}
\right.
\end{equation*}
$$

Remember from previous notebook that ```(x>0)``` translates into the boolean ```True```. However, Python can treat ```True``` as the number ```1```. Hence, the operation ```(x > 0)*x``` is interpreted as a multiplication of two nubers rather than a multiplication of a boolean and a number.

## ```zip()```

While ```map()``` works on one list (tuple), we sometimes want to compare two lists (tuples) or perform some operation that involves more than one list (tuple).

````python
midterm1 = [68, 74, 89, 81, 77, 76]
midterm2 = [71, 80, 88, 60, 90, 81]
````

````python
zip(midterm1,midterm2)  #not so usefull
````

````python
list(zip(midterm1,midterm2))
````

The function ```zip()``` inside ```list()``` creates a list of pairs taking one element from the first list and the matching element from the second list. (Note: we could write ```tuple(zip(midterm1,midterm2))``` to get a tuple of pairs rather than a list. Try it.)

````python
change = []  #this is an empty list that we want to fill with the difference in score between the two midterms
for x,y in zip(midterm1,midterm2):
    #compute the difference:
    change.append(x-y)  #the change.append syntax means the append() function is applied to the list change

print(change)
````

## ```filter()```

As the name of the function sugests, we sometimes want just a subset of the elements in a certain list (or tuple). We need to have a function that returns ```True``` or ```False``` and 

````python
def finalIsBest(m1,m2,f):
    return (f>m1) & (f>m2)  # & means that both conditions must be true
````

````python
#this is not going to work because filter() expects two arguments only: (function,list)

#try it by removing the # from the next line:
#list(filter(finalIsBest, midterm1,midterm2,final))  
````

````python
# step 1: create a list of tuples of grades
grades = list(zip(midterm1,midterm2,final))

print(grades)
````

````python
# step 2: write a function that works on a tuple (i.e. has one input)

def isFinalBest(grades):  #notice I slightly changes the name of the function
    return (grades[2]>grades[0]) & (grades[2]>grades[1])

# Notice that in Python the index starts from zero!
````

````python
list(filter(isFinalBest,grades))
````
