**Definition:** A _mutable object_ is an object whose state or value can be changed after it is created. This means that the object's data can be modified, added, or removed without changing its identity. Mutable objects can be altered in place, and any references to the object will reflect these modifications.

To better understand the way _Python_ handles mutable and immutable objects we will use the `id()` function.

The id() function in Python returns the unique identifier of an object. This identifier is an integer that represents the _memory address_ where the object is stored in the computer's memory. The id() function allows you to obtain this unique identifier for any object in Python.

Syntax:
```Python
id(name_of_object)
```

## Mutable objects

Some examples of mutable objects in Python include lists, dictionaries, sets, and byte arrays.

if you have a list object in Python, you can change its elements, add new elements, or remove existing elements without creating a new list object. Here's an example:

````python
my_list = [1, 2, 3]
print(my_list)
id(my_list)
````

````python
my_list.append(4)
print(my_list)
id(my_list)
````

````python
my_list[0] = 5
print(my_list)
id(my_list)
````

As you can see, the variable `my_list` points to the same _memory address_ before and after it is modified. This is true because a list is a mutable object in Python.

## Immutable objects

An immutable object is an object whose state or value cannot be changed after it is created. Once an immutable object is assigned a value, it cannot be modified, and any attempt to do so will result in the creation of a new object with the desired changes. Immutable objects in Python are inherently safe from unintended modifications.

Examples: numbers, string, tuppels.

````python
x = 5
print(id(x))
````

````python
x = 6
print(id(x))
````

The id changed! 

In other words, we didn't really modify the object named `x`, we destroyed the first object and created a new one with the same name.

Another example with strings:

````python
my_string = "Hello"
print(id(my_string))  

my_string += " World"
print(id(my_string)) 
````

````python
i = 8
i = i+1
print(i)

i -=1
print(i)
````

Again, not the same `id`.

## Converting mutable to immutable

Making an object immutable will prevent us from changing it by mistake.

Here is an example:

````python
numbers = [1, 2, 2, 3, 4, 4, 5]
frozen_numbers = frozenset(numbers)
print(frozen_numbers)  
````

The above object is not easy to work with and duplicate elements have been eliminated. Another possibility is to convert a list to a tuple.

````python
tuple_numbers = tuple(numbers)
print(tuple_numbers)
````
