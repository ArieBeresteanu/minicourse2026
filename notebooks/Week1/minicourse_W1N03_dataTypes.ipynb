## Basic types

Python has the following data types built-in by default, in these categories:

**Type we will use:**

- Text Type:	`str`
-  Numeric Types:	`int`, `float`
- Sequence Types:	`list`, `tuple`, `range`
- Mapping Type:	`dict`
- Boolean Type:	`bool`
- None Type:	`None`

**Adittional types (we'll not use these):**

-  Numeric Types:	complex
- Set Types:	`set`, frozenset
- Binary Types:	bytes, bytearray, memoryview  



_Dictionaries_ will be covered in another notebook!

## Advanced Topics

### Compare value AND type

Python can see that ```4.0 == 4``` even if they are not of the same type. You can compare **BOTH** value and type with the ```type``` function:

````python
(y == 4) and (type(y) == type(4))

# Note y=4.0 which is of a different type than 4
````
