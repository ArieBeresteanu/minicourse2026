הֲזָחָה
	
In Python properly indenting your code is **crucial**!!! 

Indentation matters in

- Functions
- For/while loops
- If/else sections

## Functions

Example:
```
def cubeIt(x)
return pow(x,3)
```

The above code will **NOT** work!

```Python
def cubeIt(x):
    return pow(x,3)
```

This is the right syntax!

You need the following two ingredients:

- A collon ```:``` at the end of the function name and inputs declaration
- Indent everything below it (uniformly one level)

**Notice:** There is no closing ```end``` statement like in many other languages.

````python
def cubeIt(x):
    return pow(x,3)
````

````python
cubeIt(2.3)
````

## For loops

Similarly to function declaration you need a ```:``` at the end of the for loop first line and then indentation (and no ```end``` statement).

````python
for x in [1,4,9,16,-25]:
    print(pow(x,0.5))
````

## If Elif Else

Same as above but now a ```:``` and indentation are needed for each section of the if/elif/else statement.

````python
if x >0:
    print("It's positive") #note the use of " versus '
elif x<0:
    print("It's negative")
else:
    print ("It's zero")
````

````python
print(x)
````

## Nested Indentation

The above rules on indentation imply to _nested_ statements that requier indentation.

````python
def sumIt(x):
    #first level indentation
    s = 0
    
    for i in range(int(x)):
        # second level indentation
        s += i
    
    print("The sum of all numbers from 1 to",x,"is",s)
    if s > 100:
        # second level indentation still
        print ("It's a big sum!")
        if s < 1000:
            # Third level of indentation
            print("But it's not huge!")
    else:
        print("It's a modest sum.")
    
    return s
    
````

**Note:** Explaining the function `range()` will come in a later notebook.

````python
sumIt(20);  #add a semi-colon at the end of the function call
````
